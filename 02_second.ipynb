#### File operations

```python
# Load the file and find the average of the numbers
f = open('mydata.txt')
content = f.read()
print(content)
```

```python
from nltk import word_tokenize
import string
```

```python
a = word_tokenize(content)
nums = []
for word in a:
    if not word.isalpha() and word not in string.punctuation :
        nums.append(float(word))
print(nums)
```

```python
sum(nums) / len(nums)
```

#### Multi Word Expression Tokenizer

```python
from nltk.tokenize import MWETokenizer
```

```python
nsent = 'Chhatrapati Shivaji Maharaj was founder of Maratha Empire'
```

```python
word_tokenize(nsent)
```

```python
mwe = MWETokenizer(separator = ' ') #changing the separator to space (' '), by default it is underscore (_)
```

```python
mwe.add_mwe(('Chhatrapati','Shivaji'))
```

```python
mwe.tokenize(word_tokenize(nsent))
```

#### Custom Tokenizor

```python
import re
```

```python
def custom_tokenizer(text):
    return re.split(r",.;?!\s]+",text)

text = "This is some text with punctuation > Let's tokenize it. Is it ok?"
```

```python
custom_tokenizer(text)
```

```python
class CommaTokenizer:
    def tokenize(self, text):
        return re.split(r"[,\s]+", text)

tk = CommaTokenizer()
tk.tokenize(text)
```
