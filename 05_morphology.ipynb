```python
words = ['cats','computing','association','programming','leadership','miscommunication','identifiable','psychologically','engineering','tabular', 'realistic','colorfulness']
```

```python
from polyglot.text import Word
```

```python
for w in words:
    w = Word(w, language = 'en')
    print(w, '\t', w.morphemes)
```

```python
mwords = ['महाद्वीप','वायुमंडल','महाराजा']
for word in mwords:
    w = Word(word, language = 'mr')
    print(word, '\t\t', w.morphemes)
```
