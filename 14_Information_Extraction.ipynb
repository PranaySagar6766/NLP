#### Find top 10 words occured frequently in this article

```python
from nltk.tokenize import word_tokenize
```

```python
tokens = word_tokenize(text)
```

```python
from nltk import FreqDist
```

```python
freq = FreqDist(tokens)
```

```python
freq.plot(10)
```

```python
from nltk.corpus import stopwords
```

```python
tokens1 = [token for token in tokens
           if token.isalpha() and token not in stopwords.words('english')
          ]
```

```python
freq = FreqDist(tokens1)
freq.plot(10)
```

```python
from nltk.corpus import stopwords
```

```python
tokens1 = [token.lower() for token in tokens
           if token.isalpha() and token.lower()
           not in stopwords.words('english')]
```

```python
from nltk import FreqDist
```

```python
freq = FreqDist
```

```python
freq = FreqDist(tokens1)
```

```python
freq.plot(10)
```

```python
freq.most_common(20)
```

```python
swords = stopwords.words('english') + ['retrieved', 'archived', 'edit','b','original']
```

```python
url = 'https://en.wikipedia.org/wiki/Saturn'
response = requests.get(url, headers = headers)
soup = BeautifulSoup(response.text, 'html.parser')
text = soup.get_text()
tokens = word_tokenize(text)
tokens1 = [token.lower() for token in tokens
           if token.isalpha() and token.lower()
           not in swords]

freq = FreqDist(tokens1)
freq.plot(20)
```
