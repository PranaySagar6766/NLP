```python
def load_doc(filename):
    f = open(filename)
    text = f.read()
    f.close()
    return text
```

```python
from nltk.tokenize import word_tokenize
swords = stopwords.words('english')
```

```python
#Function to clean the documents
def clean_doc(doc):
    tokens = word_tokenize(doc)
    tokens1 = [token for token in tokens if token.isalpha()]
    tokens2 = [token for token in tokens1 if token not in swords]
    tokens3 = [token for token in tokens2 if len(token) > 1]
    return tokens3
```

```python
# load the file, clean the data and return the string
def doc_to_line(filename):
    doc = load_doc(filename)
    tokens = clean_doc(doc)
    
    return ' '.join(tokens)
```

```python
def process_test(directory):
    documents = []
    for filename in os.listdir(directory):
        if filename.startswith('cv9'):
            path = directory + '/' + filename
            docs = load_doc(path)
            tokens = clean_doc(docs)
            documents.append(tokens)
    return documents
```

```python
def process_docs(directory, is_train):
    documents = []
    for filename in os.listdir(directory):
        if is_train and filename.startswith('cv9'):
            continue
        if not is_train and not filename.startswith('cv9'):
            continue
        path = directory + '/' + filename
        docs = load_doc(path)
        tokens = clean_doc(docs)
        documents.append(tokens)
    return documents
```

```python
def load_data(is_train):
    neg = process_docs('E:/datasets/datasets/review_polarity/txt_sentoken/neg',is_train)
    pos = process_docs('E:/datasets/datasets/review_polarity/txt_sentoken/pos', is_train)
    docs = neg + pos
    labels = [0 for i in range(len(neg))] + [1 for i in range(len(pos))]
    return docs, labels
```

#### data preparation

```python
def create_tokenizer(lines):
    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(lines)
    return tokenizer
```

```python
tokenizer = create_tokenizer(train_data)
```

```python
len(tokenizer.word_index)
```

```python
x_train = tokenizer.texts_to_matrix(train_data)
x_test = tokenizer.texts_to_matrix(test_data)
```

```python
x_train.shape
```

```python
def encode_docs(tokenizer, max_length, docs):
    encoded = tokenizer.texts_to_sequences(docs)
    padded = pad_sequences(encoded, maxlen = max_length, padding = 'post')
    return padded
```

```python
max_length = max([len(sent) for sent in train_data])
```

```python
max_length
```

```python
x_train = encode_docs(tokenizer, max_length, train_data)
x_test = encode_docs(tokenizer, max_length, test_data)
```

```python
x_train.shape, x_test.shape
```

#### Define the model

```python
from keras.models import Sequential
from keras.layers import Input, Dense, Conv1D, MaxPool1D, Flatten, Embedding
from keras.utils import to_categorical, plot_model
```

```python
# create the model
model = Sequential()

# Embedding layer
model.add(Embedding(input_dim = len(vocab_list), output_dim = 128, input_length = max_length))

#Convolution layer
model.add(Conv1D(filters = 16, kernel_size =5, activation = 'relu'))
model.add(MaxPool1D(pool_size = 2))

#Flatter Layer
model.add(Flatten())

#hidden layer
model.add(Dense(16, activation = 'relu'))

#output layer
model.add(Dense(1, activation = 'sigmoid'))
```

```python
# Compile the model
model.compile(loss = 'binary_crossentropy', optimizer = 'adam',
             metrics = ['accuracy'])
```

#### Train the model

```python
model.fit(x_train, np.array(train_labels), batch_size = 5, epochs = 10)
```

#### Prediction of unknown data


```python
review1 = 'Best movie ever! It was great, I will definitely recommend it'
review2 = 'Completely money wasting experience. One of the worst movie ever.'
```

```python
def predict_sentiment(review):
    tokens = clean_doc(review)
    tokens = ' '.join(tokens)
    encoded = tokenizer.texts_to_matrix([tokens])
    yhat = model.predict(encoded)
    prob = yhat[0, 0]
    if prob > 0.5:
        return 'Postive', prob
    return 'Negative', 1-prob
    
```
