```python
# load doc in memory

def load_doc(filename):
  f= open(filename)
  text = f.read()
  f.close()
  return text
```

```python
from  nltk.tokenize import word_tokenize
swords = stopwords.words('english')
```

```python
#Function to clean document

def clean_doc(doc):
  tokens = word_tokenize(doc)
  tokens1 = [token for token in tokens if token.isalpha()]
  tokens2 = [token for token in tokens1 if token.lower() not in swords]
  tokens3 = [token for token in tokens2 if len(token)>1]
  return tokens3
```

```python
d = load_doc('/content/cv004_12641.txt')
clean_doc(d)
```

```python
# load the file, clean the data and return the string

def doc_to_line(filename):
  doc = load_doc(filename)
  tokens = clean_doc(doc)
  return ' '.join(tokens)
```

```python
def process_docs(directory, is_train):
  documents = []
  for filename in os.listdir(directory):
    if is_train and filename.startswith('cv9'):
      continue
    if not is_train and not filename.startswith('cv9'):
      continue
    path = directory + '/' + filename
    docs = load_doc(path)
    tokens = clean_doc(docs)
    documents.append(tokens)
  return documents
```

```python
def load_data(is_train):
  neg=process_docs('/content/drive/MyDrive/Dataset/review_polarity-20260603T030652Z-3-001/review_polarity/txt_sentoken/neg',is_train)
  pos=process_docs('/content/drive/MyDrive/Dataset/review_polarity-20260603T030652Z-3-001/review_polarity/txt_sentoken/pos',is_train)
  docs = neg + pos
  # print(docs)
  labels = [0 for i in range(len(neg))] + [1 for i in range(len(pos))]
  # print( labels)
  return docs, labels
```

```python
train_data, train_labels=load_data(True)
```

```python
test_data,test_labels=load_data(False)
```

```python
## Data preprocessing
```

```python
def create_tokenizer(lines):
  tokenizer=Tokenizer()
  tokenizer.fit_on_texts(lines)
  return tokenizer
```

```python
tokenizer=create_tokenizer(train_data)
```

```python
len(tokenizer.word_index)
```

```python
x_train = tokenizer.texts_to_matrix(train_data)
x_test=tokenizer.texts_to_matrix(test_data)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
def encode_docs(tokenizer,max_length,docs):
  encoded=tokenizer.texts_to_sequences(docs)
  padded=pad_sequences(encoded,maxlen=max_length,padding='post')
  return padded
```

```python
max_length=max([len(sent) for sent in train_data])
max_length
```

```python
x_train=encode_docs(tokenizer,max_length,train_data)
x_test=encode_docs(tokenizer,max_length,test_data)
```

```python
x_train.shape,x_test.shape
```

```python
from keras.models import Sequential
from keras.layers import Input,Dense,Conv1D,MaxPool1D,Flatten,Embedding
from keras.utils import to_categorical,plot_model
```

```python
# create the model
model = Sequential()

model.add(Embedding(input_dim=len(vocab_list),output_dim=128,input_length=max_length))
model.add(Conv1D(filters=16,kernel_size=5,activation='relu'))
model.add(MaxPool1D(pool_size=2))
model.add(Flatten())
model.add(Dense(16,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
len(vocab_list)
```

```python
model.summary()
```

```python
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
#fitting the model
model.fit(x_train,np.array(train_labels),epochs=10,batch_size=5,validation_data=(x_test,np.array(test_labels)))
```

```python
y_pred_porba=model.predict(x_test)
y_pred=np.where(y_pred_porba>0.5,1,0)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,ConfusionMatrixDisplay
```

```python
accuracy_score(test_labels,y_pred)
```

```python
ConfusionMatrixDisplay.from_predictions(test_labels,y_pred)
```
