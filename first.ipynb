```python
import nltk
```

```python
sent = 'They told that their ages are 24 31 and 29 respectively'
#find averages of ages mentioned in this sentence
```

```python
ages = [int(i) for i in sent.split() if i.isdigit()]
print(ages)
```

```python
sum(ages)/len(ages)
```

```python
ages = []
for i in sent.split():
    if i.isdigit():
        ages.append(int(i))
print(sum(ages)/len(ages))
```

#### Word and Sentence Tokenizer

```python
#import the tokenizer function
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
sent = '''Hello friends!
How are you?
Welcome to the world of Python Programmming.'''
```

```python
print(sent)
```

```python
sents = sent_tokenize(sent)
```

```python
sents
```

```python
sents[0]
```

```python
sents[1]
```

```python
sents[2]
```

```python
words = word_tokenize(sent)
words
```

```python
#Among all the tokens, find % of punctuation symbols in it
```

```python
import string
```

```python
p_count = 0
for i in words:
    if i in string.punctuation:
        p_count+=1
```

```python
p_count
```

```python
p_count/len(word_tokenize(sent)) *100
```

```python
new_str = ''
for char in sent:
    if char not in string.punctuation:
        new_str += char
```

```python
print(new_str)
```
